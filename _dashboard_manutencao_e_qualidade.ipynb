```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Definindo semente de dados
np.random.seed(101)

equipamentos = ['Torno CNC', 'Robô Solda', 'Prensa 500T']
turnos = ['Diurno', 'Noturno']

# 180 registros de manutenção
reg_equipamentos = np.random.choice(equipamentos, size=180)
reg_turnos = np.random.choice(turnos, size=180)

# Métricas simuladas
custo_reparo = np.random.normal(loc=7.5, scale=1.8, size=180) # em R$ mil
tempo_reparo = np.random.normal(loc=45, scale=12, size=180) # em minutos
```

```python
# 1. Grafico 1 (Lado Esquerdo - axes[0]): Custo Médio de reparo (R$ mil) (Barplot)
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# 2. Desenhando o Barplot
sns.barplot(
    x=reg_equipamentos,
    y=custo_reparo,
    hue=reg_turnos,
    palette='Set1',
    ax=axes[0]
)

axes[0].set_title('Custo Médio de Reparo por Equipamento e Turno', fontsize=12)
axes[0].set_xlabel('Equipamento', fontsize=10)
axes[0].set_ylabel('Custo Médio (R$ mil)', fontsize=10)
axes[0].grid(axis='y', linestyle='--', alpha=0.5)

# Grafico 2 (Lado Direito - axes[1]): Distribuicão do Tempo de Reparo (violinplot)

# 3. Desenhando o Violin Plot
sns.violinplot(
    x=reg_equipamentos,
    y=tempo_reparo,
    hue=reg_turnos,
    palette='Set2',
    split=True,
    ax=axes[1]
)


# 4. Personalizacão Executiva Padronizada
axes[1].set_title('Distribuicão do Tempo de Reparo (Diurno vs Noturno)', fontsize=12)
axes[1].set_xlabel('Equipamento', fontsize=10)
axes[1].set_ylabel('Tempo de Reparo (minutos)', fontsize=10)
axes[1].grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
